# Engenharia de Features

Nesse notebook iremos fazer a engenharia de features do nosoo dataset.

## Importando as Bibliotecas

Primeiro vamos importar nossas bibliotecas mais usadas.

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## Carregando o Dataset

Vamos carregar o nosso dataset.

```python
df = pd.read_parquet("../data/processed/UCMF_analysed.parquet")
```

## Features Temporais

Inicialmente vamos tratar das features temporais. Como os modelos de ML na maioria não conseguem lidar com dados temporais, precisamos transformar a informação dessa feature em features categóricas ou numéricas. Inicialmente vamos tranformar em outras features numéricas.

```python
df["nascimento_dia"] = df["nascimento"].dt.day
df["nascimento_mes"] = df["nascimento"].dt.month
df["nascimento_ano"] = df["nascimento"].dt.year

df["atendimento_dia"] = df["atendimento"].dt.day
df["atendimento_mes"] = df["atendimento"].dt.month
df["atendimento_ano"] = df["atendimento"].dt.year

df["dias_atendimento"] = df["atendimento"] - df["nascimento"]
df["dias_atendimento"] = df["dias_atendimento"].dt.days

df = df.drop(["nascimento", "atendimento"], axis="columns")
```

## Salvando as Mudanças

Agora vamos salvar as mudanças feitas.

```python
df.to_parquet("../data/processed/UCMF_processed.parquet")
```
